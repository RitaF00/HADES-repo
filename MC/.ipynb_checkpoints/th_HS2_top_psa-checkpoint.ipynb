# Defining the physics

```python
detector = "V02162B"
measure = "th_HS2_lat_psa"
```

# GDLM file generation

```python
import dbetto
import pygeomhades
import pygeomtools
from pyg4ometry import gdml
```

```python
metadata_path = "/global/u2/r/ritaferi/HADES_DATA/hades-metadata"
db = dbetto.TextDB(metadata_path)

configuration = (
    db.hardware.configuration[detector]
    ["c1"]
    [measure]
    ["run0001"]
)

reg = pygeomhades.core.construct(configuration)
# visualise
#pygeomtools.viewer.visualize(reg)
```

```python


output_file = (
    f"/global/u2/r/ritaferi/HADES_DATA/MC/"   # directory dove salvo
    f"{detector}_{measure}_run0001.gdml"      # nome del file
)

writer = gdml.Writer()
writer.addDetector(reg)
writer.write(output_file)

print(f"File GDML creato: {output_file}")
```
