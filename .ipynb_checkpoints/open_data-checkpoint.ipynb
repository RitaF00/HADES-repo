```python
plt.figure(figsize=(8, 4))

labels = []
values = []
errors = []

for v in par['results']['qc'].keys():
    labels.append(v)
    values.append(par['results']['qc'][v]['sf_cal'])
    errors.append(par['results']['qc'][v]['sf_cal_err'])

plt.bar(
    labels,
    values,
    yerr=errors,
    color = 'deepskyblue',
    capsize=4
)

plt.ylim(97,100.5)

plt.ylabel("SF (%)")
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```
